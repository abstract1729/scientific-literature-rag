### Loading Libraries and Packages

```python
from pathlib import Path
import sys

# Project root = two levels above experiments/ingestion/
PROJECT_ROOT = Path.cwd().resolve().parents[1]

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.ingestion.pymupdf_parser import PyMuPDFParser

# Select a paper
pdf_path = (PROJECT_ROOT/ "data"/ "raw"/ "papers"/ "1706.03762.pdf")
```

### PyMuPDF Parser

```python
# Parse
parser = PyMuPDFParser()
document = parser.parse(pdf_path)

# Inspect
print("Paper ID:", document.paper_id)
print("Pages:", len(document.pages))
print("Characters:", len(document.full_text))

print("\n--- First page ---\n")
print(document.pages[0].text[:300])
```

### Parsing analysis

```python
print("=" * 70)
print("PARSER OUTPUT SUMMARY")
print("=" * 70)

print("Paper ID       :", document.paper_id)
print("Pages          :", len(document.pages))
print("Characters     :", len(document.full_text))
print("Words          :", len(document.full_text.split()))

print("\nMetadata:")
for key, value in document.metadata.items():
    print(f"  {key}: {type(value).__name__}")
```

#### First and the last characters

```python
print("FIRST 1000 CHARACTERS")
print("=" * 70)
print(document.full_text[:1000])

print("\n\nLAST 1000 CHARACTERS")
print("=" * 70)
print(document.full_text[-1000:])
```

#### Page-wise analysis

```python
## Showing first and last part of each page.

for page in document.pages[:4]:
    print("=" * 70)
    print(f"PAGE {page.page_number}")
    print("=" * 70)
    print("FIRST PART OF THE PAGE")
    print(page.text[:500])

    print("\nLAST PART OF THE PAGE")
    print(page.text[-500:])
```

### Saving the file

```python
from pathlib import Path

output_dir = (PROJECT_ROOT/ "results"/ "ingestion"/ "parser_outputs"/ "pymupdf")
output_dir.mkdir(parents=True, exist_ok=True)
output_file = output_dir / f"{document.paper_id}.md"
output_file.write_text(
    document.full_text,
    encoding="utf-8",
)

print(f"Saved to: {output_file}")
```
